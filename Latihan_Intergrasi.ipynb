```python
# Import library yang dibutuhkan
from skimage import io
import cv2 as cv
from google.colab.patches import cv2_imshow # Hapus baris ini jika tidak menggunakan Google Colab

# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://picsum.photos/800/600.jpg"]

# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
    image = io.imread(url)                          # read image
    image = cv.resize(image, (0,0), fx=0.5, fy=0.5) # resize image to half size
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)  # convert color to RGB
    final_frame = cv.hconcat((image, image_2))      # concatenate image
    cv2_imshow(final_frame)                         # show image
    print('\n')
```

```python
# Import library yang dibutuhkan
import cv2 as cv
import numpy as np
from skimage import io
from google.colab.patches import cv2_imshow

# Menggunakan URL gambar yang aman dan stabil (dari contoh kode Anda sebelumnya)
url = "https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg"

# Membaca gambar
image = io.imread(url)
image = cv.cvtColor(image, cv.COLOR_RGB2BGR) # Konversi ke BGR untuk OpenCV

# Perkecil ukuran gambar sebesar 50% agar tampilan tidak terlalu besar di layar
image = cv.resize(image, (0, 0), fx=0.5, fy=0.5)
tinggi, lebar = image.shape[:2]

# Tugas 2A: Garis Vertikal
img_vertikal = image.copy()
x_pos = int(lebar * 0.75) # Posisi garis di 3/4 lebar gambar
cv.line(img_vertikal, (x_pos, 0), (x_pos, tinggi), (255, 255, 255), 2)
print("Tugas 2A: Garis Vertikal")
cv2_imshow(cv.hconcat((image, img_vertikal)))
print('\n')

# Tugas 2B: Garis Menyilang (Diagonal)
img_diagonal = image.copy()
cv.line(img_diagonal, (0, 0), (lebar, tinggi), (255, 255, 255), 2)
cv.line(img_diagonal, (lebar, 0), (0, tinggi), (255, 255, 255), 2)
print("Tugas 2B: Garis Menyilang (Diagonal)")
cv2_imshow(cv.hconcat((image, img_diagonal)))
print('\n')

# Tugas 3: Garis Horizontal di Tengah
img_horizontal = image.copy()
y_tengah = int(tinggi / 2)
x_mulai = int(lebar * 0.3)
x_selesai = int(lebar * 0.7)
cv.line(img_horizontal, (x_mulai, y_tengah), (x_selesai, y_tengah), (255, 255, 255), 2)
print("Tugas 3: Garis Horizontal Berwarna Putih di Bagian Tengah")
cv2_imshow(cv.hconcat((image, img_horizontal)))
print('\n')

# Tugas 4: Kotak Pixel Warna Putih
img_kotak = image.copy()
img_kotak[100:200, 100:200] = [255, 255, 255]

print("Tugas 4: Kotak Pixel Putih di Sembarang Tempat")
cv2_imshow(cv.hconcat((image, img_kotak)))
```
