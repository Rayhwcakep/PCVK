```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# 1. Definisikan koordinat titik sudut (vertices)
vertices = np.array([[100, 300], [200, 200], [400, 300], [200, 400]], dtype=np.int32)

# 2. Reshape array menjadi format yang dibutuhkan polylines (N x 1 x 2)
# (Angka 2 di sini artinya koordinat X dan Y, bukan channel warna RGB!)
pts = vertices.reshape((-1, 1, 2))

# 3. Buat gambar kosong dan gambar poligonnya
black_img2 = np.zeros(shape=(512, 512, 3), dtype=np.uint8)
cv.polylines(black_img2, [pts], isClosed=True, color=(255, 0, 0), thickness=5)

# 4. Tampilkan hasil
plt.imshow(black_img2)
plt.show()
```
