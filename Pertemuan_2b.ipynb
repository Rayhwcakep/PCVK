```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Membaca foto aktivitas (pastikan path file sudah benar)
path_gambar = "/content/KTM CONTOH.jpeg" # Atau ganti dengan path foto aktivitas Anda
foto_asli = cv2.imread(path_gambar)
foto_rgb = cv2.cvtColor(foto_asli, cv2.COLOR_BGR2RGB)

# ==========================================
# TUGAS 1: Kombinasi Kanal Warna (Merah-Biru & Hijau-Biru)
# ==========================================
# Membuat duplikat citra untuk masing-masing kombinasi
citra_mb = foto_rgb.copy()
citra_hb = foto_rgb.copy()

# Kanal yang tidak dipakai dinolkan (Indeks kanal: 0=Merah, 1=Hijau, 2=Biru di format RGB)
citra_mb[:, :, 1] = 0  # Matikan kanal Hijau (Red & Blue saja)
citra_hb[:, :, 0] = 0  # Matikan kanal Merah (Green & Blue saja)

# Menampilkan hasil tugas 1 berdampingan
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(citra_mb)
plt.title("Kombinasi Kanal Merah-Biru (Tanpa Hijau)")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(citra_hb)
plt.title("Kombinasi Kanal Hijau-Biru (Tanpa Merah)")
plt.axis('off')

plt.tight_layout()
plt.show()

# ==========================================
# TUGAS 2: Pencerminan Citra (Flip Vertikal, Horizontal, Keduanya)
# ==========================================
flip_horizontal = cv2.flip(foto_rgb, 1)  # flipCode 1 = horizontal (sumbu Y)
flip_vertikal = cv2.flip(foto_rgb, 0)    # flipCode 0 = vertikal (sumbu X)
flip_keduanya = cv2.flip(foto_rgb, -1)   # flipCode -1 = keduanya (vertikal & horizontal)

# Menampilkan hasil tugas 2 dalam satu figure berisi 3 subplot
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(flip_horizontal)
plt.title("Pencerminan Horizontal")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(flip_vertikal)
plt.title("Pencerminan Vertikal")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(flip_keduanya)
plt.title("Pencerminan Keduanya")
plt.axis('off')

plt.tight_layout()
plt.show()

# ==========================================
# TUGAS 3: Menambahkan Rectangle, Circle, dan Teks Nama
# ==========================================
# Karena OpenCV menggunakan format BGR untuk menggambar teks/bentuk, kita buat salinan dari format BGR asli
foto_anotasi = foto_asli.copy()

# 1. Menggambar Rectangle di area wajah/badan (Sesuaikan koordinat pt1 dan pt2)
cv2.rectangle(foto_anotasi, pt1=(140, 480), pt2=(290, 750), color=(0, 255, 0), thickness=3)

# 2. Menggambar Circle di dekat area wajah
cv2.circle(foto_anotasi, center=(215, 580), radius=60, color=(0, 0, 255), thickness=3)

# 3. Menambahkan Teks Nama Anda
teks = "Ahmad Kevin"
lokasi_teks = (100, 440) # Posisi (X, Y)
jenis_font = cv2.FONT_HERSHEY_SIMPLEX
skala_font = 1.0
warna_teks = (255, 0, 0) # Format BGR (Biru)
ketebalan_teks = 2

cv2.putText(foto_anotasi, teks, lokasi_teks, jenis_font, skala_font, warna_teks, ketebalan_teks)

# Konversi kembali ke RGB untuk ditampilkan dengan Matplotlib
foto_anotasi_rgb = cv2.cvtColor(foto_anotasi, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 8))
plt.imshow(foto_anotasi_rgb)
plt.title("Anotasi: Rectangle, Circle, dan Teks Nama")
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow  # Import penampil gambar khusus Colab

# 1. Baca citra grayscale
img_original = cv2.imread('/content/KTM CONTOH.jpeg', cv2.IMREAD_GRAYSCALE)

# Cek apakah file berhasil dibaca
if img_original is None:
    print("Gambar nggak ketemu! Pastikan nama dan path filenya udah bener.")
else:
    # 2. Hitung rata-rata awal
    mean_original = np.mean(img_original)

    # 3. Downsampling (50%)
    h, w = img_original.shape
    img_small = cv2.resize(img_original, (w // 2, h // 2), interpolation=cv2.INTER_AREA)

    # 4. Upsampling kembali ke ukuran awal
    img_restored = cv2.resize(img_small, (w, h), interpolation=cv2.INTER_LINEAR)

    # 5. Hitung rata-rata akhir
    mean_restored = np.mean(img_restored)

    # 6. Output angka
    print(f"Rata-rata Intensitas Asli    : {mean_original:.4f}")
    print(f"Rata-rata Intensitas Hasil   : {mean_restored:.4f}")
    print(f"Selisih Absolute             : {abs(mean_original - mean_restored):.4f}\n")

    # 7. Tampilkan citra ke layar
    print("Citra Asli:")
    cv2_imshow(img_original)

    print("Citra Hasil Upsampling:")
    cv2_imshow(img_restored)
```
